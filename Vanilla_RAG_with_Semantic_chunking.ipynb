````python
import camelot
import pandas as pd
import os
from PIL import Image

def extract_tables_from_pdf_to_csv(pdf_path, output_directory):
    """
    Extract tables from a PDF and save each table as a separate CSV file.
    Args:
        pdf_path (str): Path to the PDF file
        output_directory (str): Directory to save the CSV files
    Returns:
        list: List of paths to the created CSV files
    """
    try:
        # Attempt to extract tables using camelot
        tables = camelot.read_pdf(pdf_path, pages='all')
        table_details = []

        for i, table in enumerate(tables):
            df = table.df
            # Convert the table to a structured format for VisionLLM
            table_data = df.to_dict(orient="records")

            # Prepare a prompt for VisionLLM to analyze the table structure
            prompt = f"Analyze the following table and provide a description of the table structure and the data contained within it:\n\n"
            for row in table_data:
                prompt += str(row) + "\n"

            # Use VisionLLM to analyze the extracted table
            # Note:  VisionLLM is a multimodal model, so it would typically take an image and text.
            #        Since we don't have the image of the PDF page, we are providing only the table data.
            #        For better accuracy, use VisionLLM with an image of the table from the PDF.
            response = vision_llm.invoke(prompt)
            table_details.append(response)

            csv_file_path = os.path.join(output_directory, f"table_{i+1}.csv")
            df.to_csv(csv_file_path, index=False)

        return table_details

    except Exception as e:
        print(f"Error processing {pdf_path}: {e}")
        return []
````

````python
from dotenv import load_dotenv
import numpy as np
import re
import os
from typing import List
from sklearn.metrics.pairwise import cosine_similarity
from pypdf import PdfReader
from qdrant_client import QdrantClient
from qdrant_client.models import VectorParams, Distance
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_qdrant import QdrantVectorStore

#%% Environment Setup
load_dotenv()
qdrant_client = QdrantClient(":memory:")
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

#%% Semantic Chunking Functions
def semantic_chunking(text: str, buffer_size: int = 1, breakpoint_percentile: int = 95) -> List[str]:
    """Split text into semantically coherent chunks using embedding distances"""
    # Split into sentences
    sentences = [s.strip() for s in re.split(r'(?<=[.?!])\s+', text) if s.strip()]
    if not sentences:
        return []

    # Create sentence windows
    sentence_windows = []
    for i in range(len(sentences)):
        window = sentences[max(0, i - buffer_size):i + buffer_size + 1]
        sentence_windows.append({
            'text': ' '.join(window),
            'index': i,
            'original': sentences[i]
        })

    # Get embeddings for windows
    window_embeddings = embeddings.embed_documents([w['text'] for w in sentence_windows])

    # Calculate distances between consecutive windows
    distances = []
    for i in range(len(sentence_windows) - 1):
        emb1 = window_embeddings[i]
        emb2 = window_embeddings[i + 1]
        distances.append(1 - cosine_similarity([emb1], [emb2])[0][0])

    # Find semantic breakpoints
    threshold = np.percentile(distances, breakpoint_percentile)
    breakpoints = [i for i, d in enumerate(distances) if d > threshold]

    # Create chunks from breakpoints
    chunks, start_idx = [], 0
    for bp in breakpoints:
        chunks.append(' '.join(sentences[start_idx:bp+1]))
        start_idx = bp + 1
    if start_idx < len(sentences):
        chunks.append(' '.join(sentences[start_idx:]))

    return chunks

#%% Document Processing
def process_pdf_documents(directory_path: str) -> List[dict]:
    """Process PDF documents with semantic chunking"""
    documents = []
    for filename in os.listdir(directory_path):
        if filename.endswith(".pdf"):
            filepath = os.path.join(directory_path, filename)
            with open(filepath, "rb") as f:
                reader = PdfReader(f)
                full_text = "\n".join([page.extract_text() for page in reader.pages])

                # csv_paths = extract_tables_from_pdf_to_csv(filepath, "tables")

                # Perform semantic chunking
                chunks = semantic_chunking(full_text, buffer_size=3, breakpoint_percentile=90)

                # Create document entries
                for chunk in chunks:
                    documents.append({
                        'text': chunk,
                        'source': filename,
                        'embedding': embeddings.embed_query(chunk)
                    })


                # Extract table details (as text) from the PDF
                # table_details = extract_tables_from_pdf_to_csv(filepath, "tables")
                # for table_text in table_details:
                #     documents.append({
                #         'text': table_text,
                #         'source': filename,
                #         'type': 'table'
                #     })

                print(f"Processed {filename} with {len(chunks)} chunks and extracted")
    return documents

#%% Vector Store Setup
def initialize_vector_store(collection_name: str = "EFH-KBASE"):
    """Initialize Qdrant vector store with proper configuration"""
    # Check if the collection exists before creating it
    if collection_name not in qdrant_client.get_collections().collections:
        qdrant_client.create_collection(
            collection_name=collection_name,
            vectors_config=VectorParams(size=384, distance=Distance.COSINE)
        )

    return QdrantVectorStore(
        client=qdrant_client,
        collection_name=collection_name,
        embedding=embeddings,
    )
````

````python
from typing_extensions import List, TypedDict
from langchain_core.documents import Document
from langchain_core.callbacks import BaseCallbackHandler
from langchain_core.outputs import LLMResult
import time
from IPython.display import display, Markdown, clear_output
from langsmith.wrappers import wrap_openai
from langsmith import traceable

#%% RAG Pipeline
class State(TypedDict):
    question: str
    context: List[str]
    answer: str
    status: str

class ProgressCallbackHandler(BaseCallbackHandler):
    """Callback handler for displaying progress updates and streaming tokens."""

    def __init__(self, display_method="notebook"):
        """Initialize the callback handler.

        Args:
            display_method: How to display progress ("notebook" for Jupyter or "console" for terminal)
        """
        self.display_method = display_method
        self.full_response = ""
        self.start_time = None

    def on_llm_start(self, serialized, prompts, **kwargs):
        """Called when LLM starts processing."""
        self.start_time = time.time()
        self._update_progress(f"🧠 Thinking...", new_block=True)

    def on_llm_new_token(self, token, **kwargs):
        """Stream tokens as they're generated."""
        self.full_response += token
        if self.display_method == "notebook":
            clear_output(wait=True)
            display(Markdown(f"```\n{self.full_response}\n```"))
        else:
            print(token, end="", flush=True)

    def on_llm_end(self, response: LLMResult, **kwargs):
        """Called when LLM generation ends."""
        elapsed = time.time() - self.start_time
        self._update_progress(f"✅ Generation completed in {elapsed:.2f}s", new_block=True)

    def on_chain_start(self, serialized, inputs, **kwargs):
        """Called when a chain starts."""
        chain_type = serialized.get("name", "Chain")
        self._update_progress(f"⚙️ Starting {chain_type}...", new_block=True)

    def on_chain_end(self, outputs, **kwargs):
        """Called when a chain ends."""
        self._update_progress("✓ Chain completed", new_block=False)

    def on_retriever_start(self, serialized, query, **kwargs):
        """Called when retriever starts."""
        self._update_progress(f"🔍 Searching for relevant documents...", new_block=True)

    def on_retriever_end(self, documents, **kwargs):
        """Called when retriever ends."""
        self._update_progress(f"📄 Found {len(documents)} relevant documents", new_block=False)

    def _update_progress(self, message, new_block=False):
        """Update progress display."""
        if self.display_method == "notebook":
            if new_block:
                clear_output(wait=True)
            display(Markdown(f"**Status:** {message}"))
        else:
            print(f"\n{message}" if new_block else f" | {message}")

@traceable
def retrieve(state: State) -> dict:
    """Enhanced semantic search retrieval with diversity optimization"""
    # Initial retrieval with higher k for candidate pool
    candidates = vector_store.similarity_search_with_score(
        query=state["question"],
        k=10
    )

    # Sort by relevance score
    sorted_docs = sorted(candidates, key=lambda x: x[1])

    # Take top 5 most relevant
    results = [doc for doc, _ in sorted_docs[:5]]

    return {"context": [doc.page_content for doc in results]}

@traceable
def generate(state: State) -> dict:
    """Generate answer with enhanced factual accuracy and executive summary"""

    state["status"] = "generating"

    context_str = "\n\n".join(state["context"])
    question = state["question"]

    # Main analysis and answer generation prompt
    unified_prompt = f"""
    Context: {context_str}

    Question: {question}

    Instructions:
    1. First, analyze the context and identify all key facts, figures, statistics, and numerical data relevant to the question.
    2. For each finding, note the exact values, confidence intervals, p-values, and statistical significance.
    3. Then, synthesize this information into a clear, structured answer that:
       - Uses ONLY information present in the context
       - Presents ALL numerical data with exact precision
       - Maintains original statistical notation (e.g., [0.09 to 0.33], p=0.026)
       - Organizes information logically using headings and bullet points
       - Clearly distinguishes between significant and non-significant findings
       - Creates a concise tabular summary of key findings when appropriate
    4. Format your answer for maximum readability and practical utility

    Answer:
    """

    # Generate comprehensive answer with lower temperature for factual consistency
    initial_answer = llm.invoke(unified_prompt, temperature=0.1).content

    # Focused verification prompt
    verification_prompt = f"""
    Context: {context_str}

    Draft answer: {initial_answer}

    Task: Please verify and correct this answer focusing ONLY on factual accuracy:
    1. Check all numerical values match exactly what's in the context
    2. Verify statistical notations (p-values, confidence intervals)
    3. Ensure no information is included that isn't supported by the context
    4. Fix any inaccuracies while preserving the structure and readability

    Corrected answer:
    """

    # Final answer with verification
    verified_answer = llm.invoke(verification_prompt, temperature=0.0).content

    # Add executive summary prompt
    summary_prompt = f"""
    Based on this verified detailed answer:

    {verified_answer}

    Create a concise executive summary (3-5 bullet points maximum) that captures the most important findings and implications.
    Focus only on the highest-impact numerical results and key takeaways.
    Do not introduce any new information not present in the verified answer.

    Executive Summary:
    """

    # Generate executive summary
    executive_summary = llm.invoke(
        summary_prompt,
        temperature=0.2
    ).content

    # Combine into well-structured final output
    structured_output = f"""
    ## EXECUTIVE SUMMARY
    {executive_summary}

    ## DETAILED ANALYSIS
    {verified_answer}
    """

    state["status"] = "completed"
    return {"answer": structured_output}
````

````python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate

source_attribution_prompt = """
    You are a source detection system. Your task is to identify which specific parts of the answer are derived from which source documents.

    Source Documents:
    {sources}

    Final Answer:
    {answer}

    For each paragraph or major point in the answer, identify which source document(s) it came from.
    Format your response as follows:

    [Attribution]
    Paragraph/Point 1: Source(s) X, Y
    Paragraph/Point 2: Source Z
    ...etc.

    Only list sources that are actually used. If information appears to be synthesized from multiple sources, list all relevant sources.
    """


# Create a function to be called after answer generation
@traceable
def annotate_with_sources(state: State) -> dict:
    """Add source attribution to the generated answer."""
    # Get the sources for each context chunk
    sources_info = {}
    for i, chunk in enumerate(state["context"]):
        # Get document metadata for this chunk from vector store
        results = vector_store.similarity_search(chunk, k=1)
        if results:
            doc_source = results[0].metadata.get("source", f"Unknown Source {i}")
            sources_info[f"Chunk {i+1}"] = f"{doc_source}: {chunk[:100]}..."

    # Format sources info for the prompt
    sources_str = "\n\n".join([f"{k}: {v}" for k, v in sources_info.items()])

    # Get attribution analysis
    attribution_prompt = PromptTemplate.from_template(source_attribution_prompt)
    attribution_chain = attribution_prompt | llm | StrOutputParser()

    attribution = attribution_chain.invoke({
        "sources": sources_str,
        "answer": state["answer"]
    })

    # Combine the answer with attribution
    full_answer = f"{state['answer']}\n\n{attribution}"

    return {"answer": full_answer}
````

````python
import re

def extract_final_answer(llm_response):
    """
    Extracts the final answer from an LLM response by removing <think> tags
    and keeping only the relevant output.
    """
    # Ensure response is a string
    if not isinstance(llm_response, str):
        raise ValueError("Input must be a string.")

    # Remove <think>...</think> section
    cleaned_response = re.sub(r"<think>.*?</think>", "", llm_response, flags=re.DOTALL).strip()

    return cleaned_response
````

````python
from IPython.display import display, Markdown, HTML, clear_output
from typing import List, Dict, Any
from langchain_core.callbacks import CallbackManager
import pandas as pd
import ipywidgets as widgets
import re
import time

class ConversationRAG:
    """Interactive RAG system with conversation history, streaming, and source attribution."""

    def __init__(self, graph, vector_store, llm):
        """Initialize the conversation RAG system.

        Args:
            graph: The LangGraph for RAG
            vector_store: Vector store for retrieval
            llm: Language model for generation
        """
        self.graph = graph
        self.vector_store = vector_store
        self.llm = llm
        self.conversation_history = []
        self.sources_by_chunk = {}

    def extract_sources_from_context(self, context_chunks: List[str]) -> Dict[str, List[str]]:
        """Extract source documents from retrieved context chunks."""
        # Get documents with their metadata
        query_results = self.vector_store.similarity_search(
            query=" ".join(context_chunks[:2]),  # Use first chunks as query to find matching docs
            k=len(context_chunks)
        )

        # Create a mapping from chunk text to source document
        chunk_to_source = {}
        for chunk in context_chunks:
            # Find the closest match in the retrieved documents
            best_match = None
            best_score = 0

            for doc in query_results:
                # Simple string overlap score (can be improved with embeddings similarity)
                score = self._string_overlap_score(chunk, doc.page_content)
                if score > best_score:
                    best_score = score
                    best_match = doc

            if best_match and best_score > 0.5:  # Threshold for confidence
                source = best_match.metadata.get("source", "Unknown")
                if chunk not in chunk_to_source:
                    chunk_to_source[chunk] = []
                if source not in chunk_to_source[chunk]:
                    chunk_to_source[chunk].append(source)

        return chunk_to_source

    def _string_overlap_score(self, str1: str, str2: str) -> float:
        """Calculate simple string overlap score."""
        # Normalize strings
        str1 = str1.lower().strip()
        str2 = str2.lower().strip()

        # Get overlap count of 5-grams
        ngram_size = 5
        if len(str1) < ngram_size or len(str2) < ngram_size:
            return 0

        str1_ngrams = [str1[i:i+ngram_size] for i in range(len(str1)-ngram_size+1)]
        str2_ngrams = [str2[i:i+ngram_size] for i in range(len(str2)-ngram_size+1)]

        common = set(str1_ngrams).intersection(set(str2_ngrams))

        # Calculate Jaccard similarity
        if not str1_ngrams or not str2_ngrams:
            return 0
        return len(common) / (len(set(str1_ngrams).union(set(str2_ngrams))))

    def format_sources(self, sources_dict: Dict[str, List[str]]) -> str:
        """Format source documents for display."""
        # Collect unique sources
        all_sources = set()
        for sources in sources_dict.values():
            all_sources.update(sources)

        # Format as markdown
        if not all_sources:
            return "No sources found."

        sources_md = "### Sources\n"
        for i, source in enumerate(sorted(all_sources)):
            sources_md += f"{i+1}. {source}\n"

        return sources_md

    def update_conversation_history(self, question: str, answer: str):
        """Update conversation history with new Q&A pair."""
        self.conversation_history.append({
            "question": question,
            "answer": answer,
            "timestamp": time.time()
        })

    def generate_answer_with_streaming(self, question: str, output_area=None):
        """Generate answer with streaming output and source attribution."""
        # Initialize streaming output area if provided
        if output_area:
            with output_area:
                clear_output(wait=True)
                display(HTML("<p><b>Thinking...</b></p>"))

        # First, run retrieval to get context with history included
        initial_state = {
            "question": question,
            "context": [],
            "answer": "",
            "status": "retrieving"
        }
        # First, run retrieval to get context
        initial_state = {"question": question, "context": [], "answer": "", "status": "retrieving"}
        result = self.graph.invoke(initial_state, config)

        # Extract sources from context
        self.sources_by_chunk = self.extract_sources_from_context(result["context"])

        # Format the final answer with sources
        sources_section = self.format_sources(self.sources_by_chunk)
        clean_answer = re.sub(r"<think>.*?</think>", "", result["answer"], flags=re.DOTALL).strip()
        full_answer = f"{clean_answer}\n\n{sources_section}"

        # Update conversation history
        self.update_conversation_history(question, full_answer)

        # Display in output area if provided
        if output_area:
            with output_area:
                clear_output(wait=True)
                display(Markdown(full_answer))

        return full_answer

    def get_formatted_conversation_history(self) -> str:
        """Get formatted conversation history as markdown."""
        if not self.conversation_history:
            return "No conversation history yet."

        md = "# Conversation History\n\n"
        for i, entry in enumerate(self.conversation_history):
            md += f"## Q{i+1}: {entry['question']}\n\n"
            md += f"{entry['answer']}\n\n"
            md += "---\n\n"

        return md

    def display_interactive_interface(self):
        """Display an interactive interface using ipywidgets."""
        # Create widgets
        question_input = widgets.Textarea(
            placeholder='Enter your question here...',
            layout=widgets.Layout(width='100%', height='100px')
        )

        submit_button = widgets.Button(
            description='Ask Question',
            button_style='primary',
            layout=widgets.Layout(width='150px')
        )

        output_area = widgets.Output(
            layout=widgets.Layout(width='100%', border='1px solid #ddd', padding='10px')
        )

        history_button = widgets.Button(
            description='Show History',
            button_style='info',
            layout=widgets.Layout(width='150px')
        )

        history_area = widgets.Output(
            layout=widgets.Layout(width='100%', border='1px solid #ddd', padding='10px',
                                 max_height='500px', overflow='auto', display='none')
        )

        # Define button callbacks
        def on_submit_clicked(b):
            question = question_input.value
            if not question.strip():
                with output_area:
                    clear_output(wait=True)
                    print("Please enter a question.")
                return

            question_input.value = ""  # Clear input field
            self.generate_answer_with_streaming(question, output_area)

        def on_history_clicked(b):
            with history_area:
                clear_output(wait=True)
                if history_area.layout.display == 'none':
                    history_area.layout.display = 'block'
                    display(Markdown(self.get_formatted_conversation_history()))
                else:
                    history_area.layout.display = 'none'

        # Connect callbacks
        submit_button.on_click(on_submit_clicked)
        history_button.on_click(on_history_clicked)

        # Assemble layout
        header = widgets.HTML("<h2>RAG Conversation System</h2>")
        input_box = widgets.VBox([
            widgets.Label('Question:'),
            question_input
        ])

        button_row = widgets.HBox([submit_button, history_button])

        # Display the interface
        display(widgets.VBox([
            header,
            input_box,
            button_row,
            widgets.Label('Answer:'),
            output_area,
            history_area
        ]))

        return {
            'question_input': question_input,
            'submit_button': submit_button,
            'output_area': output_area,
            'history_button': history_button,
            'history_area': history_area
        }

# Enhanced Progress Callback for streaming
class EnhancedStreamingCallback(BaseCallbackHandler):
    """Enhanced callback handler for streaming tokens and progress updates."""

    def __init__(self, output_widget=None):
        """Initialize with optional output widget."""
        self.full_response = ""
        self.start_time = None
        self.output_widget = output_widget

    def on_llm_start(self, serialized, prompts, **kwargs):
        """Called when LLM starts generating."""
        self.start_time = time.time()
        self._display_update("🧠 Thinking...", clear=True)

    def on_llm_new_token(self, token, **kwargs):
        """Stream tokens as they're generated."""
        self.full_response += token
        self._display_update(self.full_response)

    def on_llm_end(self, response, **kwargs):
        """Called when LLM generation completes."""
        elapsed = time.time() - self.start_time
        self._display_update(f"{self.full_response}\n\n✅ Generated in {elapsed:.2f}s")

    def on_retriever_start(self, serialized, query, **kwargs):
        """Called when retriever starts."""
        self._display_update("🔍 Searching for relevant documents...", clear=True)

    def on_retriever_end(self, documents, **kwargs):
        """Called when retriever ends."""
        self._display_update(f"📄 Found {len(documents)} relevant documents")

    def _display_update(self, content, clear=False):
        """Update display with content."""
        if self.output_widget:
            with self.output_widget:
                if clear:
                    clear_output(wait=True)
                display(Markdown(content))
        else:
            if clear:
                clear_output(wait=True)
            display(Markdown(content))

````
